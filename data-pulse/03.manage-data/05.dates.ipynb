## Módulo datetime de Python

- datetime es el módulo estándar para trabajar con fechas y tiempos.

```python

from datetime import datetime, date, time, timedelta

# CREAR OBJETOS DE FECHA
hoy = date.today()
print(f"Hoy: {hoy}")  # 2026-02-22

ahora = datetime.now()
print(f"Ahora: {ahora}")  # 2026-02-22 13:27:39.123456

# CREAR MANUALMENTE
fecha_nacimiento = date(1990, 5, 15)
print(f"Nacimiento: {fecha_nacimiento}")

fecha_evento = datetime(2026, 2, 22, 14, 30, 45)
print(f"Evento: {fecha_evento}")

# PROPIEDADES
print(f"\nPropriedades de {ahora}:")
print(f"  Año: {ahora.year}")
print(f"  Mes: {ahora.month}")
print(f"  Día: {ahora.day}")
print(f"  Hora: {ahora.hour}")
print(f"  Minuto: {ahora.minute}")
print(f"  Segundo: {ahora.second}")

# ARITMÉTICA CON FECHAS
mañana = hoy + timedelta(days=1)
print(f"\nMañana: {mañana}")

proxima_semana = hoy + timedelta(weeks=1)
print(f"Próxima semana: {proxima_semana}")

hace_30_dias = hoy - timedelta(days=30)
print(f"Hace 30 días: {hace_30_dias}")

# DIFERENCIA ENTRE FECHAS
fecha1 = date(2026, 2, 22)
fecha2 = date(2026, 3, 22)
diferencia = fecha2 - fecha1
print(f"\nDiferencia: {diferencia.days} días")
      
```

## Parsear Strings a Datetime

- Los datos reales vienen como strings. Necesitas convertir a datetime.

```python

from datetime import datetime

# FORMATOS COMUNES
formatos = {
    "ISO": "2026-02-22",
    "US": "02/22/2026",
    "EU": "22/02/2026",
    "Largo": "February 22, 2026",
    "Con hora": "2026-02-22 14:30:45"
}

# PARSEAR CON strptime
def parsear_fecha(fecha_str, formato):
    """Convertir string a datetime"""
    try:
        return datetime.strptime(fecha_str, formato)
    except ValueError as e:
        return None

# Ejemplos
print("Parseando fechas:")

fecha_iso = parsear_fecha("2026-02-22", "%Y-%m-%d")
print(f"  ISO '2026-02-22' -> {fecha_iso}")

fecha_us = parsear_fecha("02/22/2026", "%m/%d/%Y")
print(f"  US '02/22/2026' -> {fecha_us}")

fecha_long = parsear_fecha("February 22, 2026", "%B %d, %Y")
print(f"  Largo 'February 22, 2026' -> {fecha_long}")

# DETECCIÓN AUTOMÁTICA DE FORMATO
def detectar_formato(fecha_str):
    """Intentar múltiples formatos"""
    formatos = [
        "%Y-%m-%d",           # ISO
        "%d/%m/%Y",           # EU
        "%m/%d/%Y",           # US
        "%Y-%m-%d %H:%M:%S",  # ISO con hora
        "%d-%m-%Y",
        "%m-%d-%Y"
    ]
    
    for fmt in formatos:
        try:
            return datetime.strptime(fecha_str, fmt), fmt
        except ValueError:
            continue
    
    return None, None

print(f"\nDetección automática:")
fechas_test = ["2026-02-22", "22/02/2026", "02-22-2026"]
for f in fechas_test:
    resultado, fmt = detectar_formato(f)
    print(f"  '{f}' -> {resultado} ({fmt})")

# MANEJAR ERRORES
def parsear_seguro(fecha_str, formato):
    """Parsear con manejo de errores"""
    if not fecha_str or not isinstance(fecha_str, str):
        return None
    
    try:
        return datetime.strptime(fecha_str.strip(), formato)
    except (ValueError, TypeError):
        return None

print(f"\nParseando con errores:")
print(f"  None: {parsear_seguro(None, '%Y-%m-%d')}")
print(f"  '': {parsear_seguro('', '%Y-%m-%d')}")
print(f"  'inválido': {parsear_seguro('inválido', '%Y-%m-%d')}")
print(f"  '2026-02-22': {parsear_seguro('2026-02-22', '%Y-%m-%d')}")
      
```

## Zonas Horarias (Timezone)

- Trabajar en múltiples zonas horarias requiere cuidado.
- [zoneinfo](https://docs.python.org/3/library/zoneinfo.html)

```python

from datetime import datetime, timezone
from zoneinfo import ZoneInfo

# ZONA HORARIA UTC (Universal Time Coordinated)
print("ZONA HORARIA UTC:")
ahora_utc = datetime.now(timezone.utc)
print(f"  Ahora en UTC: {ahora_utc}")

# CREAR FECHA CON ZONA HORARIA
fecha_madrid = datetime(2026, 2, 22, 14, 30, tzinfo=ZoneInfo("Europe/Madrid"))
print(f"\nFecha en Madrid: {fecha_madrid}")

fecha_ny = datetime(2026, 2, 22, 14, 30, tzinfo=ZoneInfo("America/New_York"))
print(f"Fecha en NY: {fecha_ny}")

# CONVERTIR ENTRE ZONAS
print(f"\nConvertir entre zonas:")
print(f"  Madrid 14:30 = {fecha_madrid.astimezone(ZoneInfo('America/New_York')).strftime('%H:%M')} en NY")
print(f"  NY 14:30 = {fecha_ny.astimezone(ZoneInfo('Europe/Madrid')).strftime('%H:%M')} en Madrid")

# COMPARAR FECHAS EN DIFERENTES ZONAS
fecha1 = datetime(2026, 2, 22, 14, 0, tzinfo=ZoneInfo("Europe/Madrid"))
fecha_1_utc = fecha1.astimezone(ZoneInfo("UTC"))
fecha2 = datetime(2026, 2, 22, 8, 0, tzinfo=ZoneInfo("America/New_York"))
fecha_2_utc = fecha2.astimezone(ZoneInfo("UTC"))

print(f"\nComparación:")
print(f"\tMadrid 14:00 vs NY 08:00")
print(f"\tMadrid {fecha_1_utc} | New York {fecha_2_utc}")
print(f"\t¿Son iguales? {fecha1 == fecha2}")  # True (misma hora UTC)

# PROBLEMAS CON ZONAS HORARIAS
print(f"\nProblem: Cambio de horario de verano")
print(f"  El 2026-03-29 en Madrid habrá cambio a horario de verano")
print(f"  Siempre almacena en UTC internamente, convierte al mostrar")
      
```

## Tips y Mejores Prácticas

- 💡 **Almacena en UTC** — Convierte a zona horaria al mostrar, nunca al almacenar.
- ⚠️ **Cuidado con horario de verano** — Usa `zoneinfo` (Python 3.9+), no offsets manuales.
- 💡 **Usa ISO 8601** — `YYYY-MM-DD` es el estándar internacional para intercambio de fechas.
- ℹ️ **Detecta formatos automáticamente** — Si tienes múltiples fuentes, intenta `%Y-%m-%d`, `%d/%m/%Y`, `%m/%d/%Y`.
- ⚠️ **Valida fechas inválidas** — `2026-02-30` o `2026-13-01` son inválidas. Siempre verifica.
- 💡 **Agrupa para análisis** — Mes, trimestre o año suelen ser más útiles que días exactos.

## Errores Comunes

### 1. Asumir que todas las fechas están en el mismo formato

> **¿Por qué ocurre?**  
Datos de múltiples fuentes pueden tener formatos diferentes: ISO, US, EU, custom.

> **Solución**  
Detecta el formato automáticamente o estandariza en la importación.

---

### 2. Comparar fechas sin normalizar zona horaria

> **¿Por qué ocurre?**  
`14:00` en Madrid ≠ `14:00` en NY. Parece igual pero es diferente tiempo absoluto.

> **Solución**  
Convierte a UTC para comparaciones. Muestra en zona local al usuario final.

---

### 3. Olvidar los problemas del año 2000 (Y2K)

> **¿Por qué ocurre?**  
Datos antiguos pueden tener años en 2 dígitos: `'95` = 1995 o 2095?

> **Solución**  
Convención común: `<30` → `20XX`, `>=30` → `19XX`.

---

### 4. No validar fechas antes de calcular

> **¿Por qué ocurre?**  
Una fecha inválida rompe cualquier cálculo. Un error silencioso es peor.

> **Solución**  
Valida rangos: año `1900-2100`, mes `1-12`, día válido según mes.

---

### 5. Usar strings para cálculos de fecha

> **¿Por qué ocurre?**  
`"2026-02-22" + 1` no funciona. Necesitas tipo `datetime`.

> **Solución**  
Parsea a `datetime`, haz cálculos, formatea a string solo al final si es necesario.
