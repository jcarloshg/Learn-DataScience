# Duplicates

```python
# DUPLICADOS EXACTOS
datos_exactos = [
    {"id": 1, "nombre": "Juan", "email": "juan@example.com"},
    {"id": 2, "nombre": "María", "email": "maria@example.com"},
    {"id": 1, "nombre": "Juan", "email": "juan@example.com"},  # Duplicado exacto
    {"id": 3, "nombre": "Pedro", "email": "pedro@example.com"}
]
```

## Find exact duplicates

```python
# find exact duplicates
def find_exact_duplicates(items):
    visited = []
    duplicates = []
    
    for i in items:
        print("\n")
        print(f"{i.items()}")
        print(f"{tuple(i.items())}")
        print(f"{sorted(tuple(i.items()))}")
        i_tuple = sorted(tuple(i.items()))
        if i_tuple in visited:
            duplicates.append(i_tuple)
        else:
            visited.append(i_tuple)

    print(f"duplicates: {duplicates}")
        
find_exact_duplicates(datos_exactos)
```

## Find partial duplicates

```python
# DUPLICADOS PARCIALES (mismo ID, datos diferentes)
datos_parciales = [
    {"id": 1, "nombre": "Juan", "email": "juan@example.com"},
    # Mismo ID, email diferente
    {"id": 1, "nombre": "Juan", "email": "juan.nuevo@example.com"},
    {"id": 2, "nombre": "María", "email": "maria@example.com"},
    # Mismo ID, nombre con tilde
    {"id": 2, "nombre": "Maria", "email": "maria@example.com"}
]

# by filed
def find_by_filed_name(items: [], field_name: str):
    visited = {}
    duplicated = []

    for idx, item in enumerate(items):
        print(f"{idx}, field_name: {field_name} - {item.get(field_name)}")
        value = item.get(field_name)
        if value in visited:
            duplicated.append(item)
            print((visited[value], idx, item))
        else:
            visited[value] = idx

    print(f"visited: {visited}")
    print(f"duplicated: {duplicated}")

    return duplicated


duplicated = find_by_filed_name(datos_parciales, "id")
print(f"size {len(duplicated)}")
print("duplicates: ")
for item in duplicated:
    print(f"item {item}")
```

```python
# DUPLICADOS PARCIALES (mismo ID, datos diferentes)
datos_parciales = [
    {"id": 1, "nombre": "Juan", "email": "juan@example.com"},
    # Mismo ID, email diferente
    {"id": 1, "nombre": "Juan", "email": "juan.nuevo@example.com"},
    {"id": 2, "nombre": "María", "email": "maria@example.com"},
    # Mismo ID, nombre con tilde
    {"id": 2, "nombre": "Maria", "email": "maria@example.com"}
]

def find_duplicates_by_multiple_field(items: [], field_names: []):
    """hola"""
    duplicates = {}
    visited = {}
    
    for idx, item in enumerate(items):
        key = tuple(item.get(c) for c in field_names)
        
        if key not in visited:
            visited[key] = item
        else:
            if key not in duplicates:
                duplicates[key] = [visited[key]]
            duplicates[key].append(item)
        
    return duplicates

multilpefileds = find_duplicates_by_multiple_field(datos_parciales, ["id", "email"])
for i in multilpefileds:
    items = multilpefileds[i]
    print(f"key: {i}")
    for item in items:
        print(f"item {item}")
    
```

## Similar duplicates

```python
# DATOS CON INCONSISTENCIAS
datos_similares = [
    {"id": 1, "nombre": "Juan García", "ciudad": "Madrid"},
    {"id": 2, "nombre": "juan garcia", "ciudad": "madrid"},  # Mayúsculas/minúsculas
    {"id": 3, "nombre": "Juan  García", "ciudad": " Madrid"},  # Espacios extra
    {"id": 4, "nombre": "Pedro López", "ciudad": "Barcelona"},
    {"id": 5, "nombre": "Pedro López", "ciudad": "Barcelona"}
]


def normalize_text(text: str) -> str:
    """Return text normalized"""
    if not isinstance(text, str):
        return str(text)
    else:
        return text.lower().strip()


def are_equals(reg_01: dict, reg_02: dict, fields: list) -> bool:
    """verify if two dicts are equals"""
    for field in fields:
        # print(f" field: {field}")
        val_01 = normalize_text(reg_01.get(field, ""))
        val_02 = normalize_text(reg_02.get(field, ""))
        # print(f"field: [{field}], val_01: [{val_01}] - val_02: [{val_02}]")
        if val_01 != val_02:
            return False
    return True


def find_duplicates_normalized(items: list, fields: list) -> list:
    """hola mano"""
    duplicates = []

    for i in range(len(items)):
        for j in range(i+1, len(items)):
            # print(f" i: {i}, j: {j}, {fields}")
            if are_equals(items[i], items[j], fields):
                duplicates.append((i, i, items[i], items[j]))

    return duplicates


duplicates = find_duplicates_normalized(datos_similares, ["nombre", "ciudad"])
# print(duplicates)
for i, j, reg_01, reg_02 in duplicates:
    print(f"    i = {i} -> {reg_01}")
    print(f"    j = {j} -> {reg_02}")
    print()
```

## fuzzy matching

```python
from difflib import SequenceMatcher

datos_nombres = [
    {"id": 1, "nombre": "Juan García López"},
    {"id": 2, "nombre": "Juan García"},
    {"id": 3, "nombre": "Juan Garcia Lopez"},
    {"id": 4, "nombre": "Pedro López"}
]


def fuzzy_str(str_01: str, str_02: str):
    return SequenceMatcher(None, str_01.lower(), str_02.lower()).ratio()


def find_similar(items: list, field: str, umbral=0.85):
    similar = []

    for i in range(len(items)):
        for j in range(i+1, len(items)):

            val_01 = str(items[i].get(field, ""))
            val_02 = str(items[j].get(field, ""))

            if val_01 != val_02:
                fuzzy_value = fuzzy_str(val_01, val_02)
                if fuzzy_value >= umbral:
                    similar.append((i, j, val_01, val_02, fuzzy_value))
    return similar


similar = find_similar(datos_nombres, "nombre", 0.8)
for i, j, val01, val02, fuzzy in similar:
    print(f"i: {i}, j: {j}, val01: [{val01}], val02: [{val02}], fuzzy: {fuzzy}")
```

## delete duplicates

```python
# DATOS CON DUPLICADOS
datos = [
    {"id": 1, "nombre": "Juan", "edad": 25, "activo": True},
    {"id": 2, "nombre": "María", "edad": 30, "activo": True},
    {"id": 1, "nombre": "Juan", "edad": 26, "activo": False},  # Duplicado de ID 1
    {"id": 3, "nombre": "Pedro", "edad": 35, "activo": True},
    {"id": 2, "nombre": "María", "edad": 30, "activo": True}   # Exactamente igual a ID 2
]

## option 1: keep the first one found
def keep_first_one_found(items: list, key:str) -> list:
    """keep the first one"""
    visited = set()
    result = []
    
    for record in items:
        value = record.get(key)
        if value not in visited:
            result.append(record)
            visited.add(value)
    
    return result

print(f"="*60)
print("keep_first_one_found")
print(f"="*60)
not_duplicated = keep_first_one_found(datos, "id")
for i in not_duplicated:
    print(i)
    

def keep_the_last(items:list, key:str)->list:
    """Mantener el último registro de cada clave"""
    
    last_found = {}
    
    for idx, register in enumerate(items):
        value = register.get(key)
        last_found[value] = idx
    
    result = []
    for i, item in enumerate(items):
        value = item.get(key)
        if i == last_found[value]:
            result.append(item)
    
    return result

print(f"="*60)
print("keep_the_last")
print(f"="*60)
not_duplicated = keep_the_last(datos, "id")
for i in not_duplicated:
    print(i)

#  ─────────────────────────────────────
# OPCIÓN 3: ELIMINAR DUPLICADOS EXACTOS (mismos valores en todos los campos)
#  ─────────────────────────────────────
def delete_exact_items(items: list)-> list:
    """delete exact items"""
    visited = []
    result = []
    
    for i in items:
        value_tuple = tuple(sorted(i.items()))
        if value_tuple not in visited:
            result.append(i)
            visited.append(value_tuple)

    return result

print(f"="*60)
print("delete_exact_items")
print(f"="*60)
not_duplicated = delete_exact_items(datos)
for i in not_duplicated:
    print(i)



#  ─────────────────────────────────────
# OPCIÓN 4: MANTENER MEJOR VERSIÓN (más completa/reciente)
#  ─────────────────────────────────────

def keep_best_version(items: list, key: str)->list:
    """keep the best version"""
    
    groups = {}
    for i in items:
        key_value = i.get(key)
        if key_value not in groups:
            groups[key_value] = []
        groups[key_value].append(i)
    
    results = []
    for key, items in groups.items():
        if len(items) == 1:
            results.append(items[0])
        else:
            best = max(items, key=lambda r: sum(1 for v in r.values() if v is not None))
            results.append(best)
            
    return results


print(f"="*60)
print("keep_best_version")
print(f"="*60)
not_duplicated = keep_best_version(datos, "id")
for i in not_duplicated:
    print(i)
```
