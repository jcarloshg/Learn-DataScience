# Funciones de Activación

ReLU, Sigmoid, Softmax, Tanh — cómo las redes neuronales obtienen su no-linealidad

## Introducción

Las funciones de activación introducen no-linealidad en las redes neuronales. Sin ellas, múltiples capas serían equivalentes a una sola transformación lineal. ReLU es max(0, x): retorna 0 para negativos y el valor para no-negativos.

### Objetivos de Aprendizaje

- Entender ReLU, Sigmoid, Softmax y Tanh
- Implementar y usar cada función de activación
- Conocer cuándo usar cada una
- Entender el problema del vanishing gradient

## ReLU (Rectified Linear Unit)

> ReLU: max(0, x). Retorna 0 para entradas negativas y el valor para no-negativas. Es la función más usada en hidden layers por su eficiencia y simplicidad.

```python
def relu(value):
    return max(0, value)

print(relu(-3))
print(relu(5))
print(relu(0))

print("\nCon NumPy:")
import numpy as np
x = np.array([-3, -1, 0, 1, 3])
print(np.maximum(0, x))
```

## Sigmoid

> Sigmoid: 1 / (1 + exp(-x)). Convierte valores a rango (0, 1). Usada históricamente en binary classification output. Sufre de vanishing gradient.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.array([-3, -1, 0, 1, 3])
print(f"x: {x}")
print(f"sigmoid(x): {sigmoid(x)}")

print("\nPropiedades:")
print("  sigmoid(0) = 0.5")
print("  sigmoid(→∞) → 1")
print("  sigmoid(→-∞) → 0")

print("\nDerivada: sigmoid(x) * (1 - sigmoid(x))")
print("  En x=0: 0.25")
print("  En x=±5: ~0")
```

## Softmax

> Softmax convierte un vector en probabilidades que suman 1. Es la generalización de sigmoid para multi-clase. Cada clase: exp(x_i) / sum(exp(x_j)).

```python
import numpy as np

def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / exp_x.sum()

logits = np.array([2.0, 1.0, 0.1])
print(f"Logits: {logits}")
print(f"Softmax: {softmax(logits)}")
print(f"Suma: {softmax(logits).sum()}")

print("\nInterpretación:")
print("  Clase 0: 66.7% de probabilidad")
print("  Clase 1: 24.4% de probabilidad")
print("  Clase 2: 8.9% de probabilidad")

print("\nRestar max evita overflow en exp()")
```

## Tanh (Tangente Hiperbólico)

> Tanh: (exp(x) - exp(-x)) / (exp(x) + exp(-x)). Convierte a rango (-1, 1). Centrado en cero, útil en hidden layers. También sufre vanishing gradient.

```python
import numpy as np

x = np.array([-3, -1, 0, 1, 3])
print(f"x: {x}")
print(f"tanh(x): {np.tanh(x)}")

print("\nPropiedades:")
print("  tanh(0) = 0")
print("  tanh(→∞) → 1")
print("  tanh(→-∞) → -1")

print("\nComparación con sigmoid:")
print("  Sigmoid: (0, 1) - no centrado")
print("  Tanh: (-1, 1) - centrado en 0")
```

## Comparación de Activaciones

> Cada activación tiene su uso. ReLU para hidden layers (computación rápida), Sigmoid para output binario, Softmax para output multi-clase, Tanh para hidden layers cuando necesitas salida centrada.

```python
import numpy as np

x = np.linspace(-5, 5, 11)
print(f"x: {x}")
print(f"\nReLU:     {np.maximum(0, x)}")
print(f"Sigmoid:  {[round(s, 3) for s in 1/(1+np.exp(-x))]}")
print(f"Tanh:     {np.round(np.tanh(x), 3)}")

print("\nCuándo usar cada una:")
uso = {
    "ReLU": "Hidden layers (default, rápido)",
    "Leaky ReLU": "Hidden layers (evita dead neurons)",
    "Sigmoid": "Output binario (0-1)",
    "Softmax": "Output multi-clase (suma 1)",
    "Tanh": "Hidden layers (centrado en 0)",
}
for act, desc in uso.items():
    print(f"  {act}: {desc}")
```

## El Problema del Vanishing Gradient

> En redes profundas, sigmoid y tanh tienen gradientes cerca de 0 en los extremos. Esto hace que el aprendizaje sea muy lento o se detenga en las primeras capas.

```python
import numpy as np

print("=== Vanishing Gradient ===")

x = np.array([0.5, 2.0, 5.0, 10.0])

sigmoid_deriv = sigmoid(x) * (1 - sigmoid(x))
tanh_deriv = 1 - np.tanh(x)**2

print(f"x:                 {x}")
print(f"sigmoid'(x):       {np.round(sigmoid_deriv, 4)}")
print(f"tanh'(x):          {np.round(tanh_deriv, 4)}")
print(f"relu' (sin relu):  {[1 if xi > 0 else 0 for xi in x]}")

print("\nProblema:")
print("  sigmoid'(5) ≈ 0.0066: gradiente casi 0")
print("  En redes profundas, este gradiente se multiplica")
print("  capa por capa → aprendizaje efectivo ≈ 0")

print("\nSolución:")
print("  Usar ReLU: gradiente es 0 (negativos) o 1 (positivos)")
print("  Skip connections: residual networks")
print("  Batch normalization")
```

## Leaky ReLU y ELU

> Leaky ReLU: max(0.01*x, x) permite gradiente pequeño para negativos, evitando "dead neurons". ELU usa exp(x)-1 para negativos.

```python
import numpy as np

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def elu(x, alpha=1.0):
    return np.where(x > 0, x, alpha * (np.exp(x) - 1))

x = np.array([-3, -1, 0, 1, 3])
print(f"x: {x}")
print(f"Leaky ReLU: {leaky_relu(x)}")
print(f"ELU:        {np.round(elu(x), 3)}")

print("\nDiferencia:")
print("  Leaky ReLU: lineal para negativos (pequeño)")
print("  ELU: exponencial para negativos, más suave")
```

## Tips y Mejores Prácticas

> ReLU es el default para hidden layers. Solo cambia si tienes razón para usar otra.

> Para output binario usa Sigmoid. Para multi-clase usa Softmax.

> Si tienes many dead neurons (muchos ceros), prueba Leaky ReLU o ELU.

> Nunca uses Sigmoid/Tanh en output si tus datos no están normalizados a esos rangos.

> Softmax requiere que ningún logit sea extremadamente grande (overflow). Resta el max antes de exp.

## Errores Comunes

### Usar Sigmoid en hidden layers

¿Por qué ocurre?
- Se usa por costumbre sin entender el vanishing gradient.

Solución
- Usa ReLU o Leaky ReLU en hidden layers.

### Softmax sin restar max

¿Por qué ocurre?
- exp(1000) causa overflow antes de dividir.

Solución
- Usa softmax(x - max(x)) para estabilidad numérica.

### Dead neurons por ReLU puro

¿Por qué ocurre?
- Si el weighted sum es negativo, ReLU salida 0 y gradiente 0. Neurona muere.

Solución
- Usa Leaky ReLU o verifica que inicialización y learning rate sean correctos.

### Mezclar activaciones incorrectamente

¿Por qué ocurre?
- Softmax espera logits sin normalizar, no pre-activados.

Solución
- Softmax va en la última capa con logits puros.
