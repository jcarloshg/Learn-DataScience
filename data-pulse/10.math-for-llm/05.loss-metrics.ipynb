# Loss Functions y Métricas

MSE, Cross-Entropy, MAE, Accuracy, Precision, Recall, F1 — cómo los modelos miden errores

## Introducción

Las funciones de pérdida miden qué tan mal está funcionando un modelo. MSE (Mean Squared Error) promedia los errores al cuadrado: ((1-1)² + (2-4)² + (3-2)²) / 3 = (0 + 4 + 1) / 3 = 1.67.

### Objetivos de Aprendizaje

- Implementar MSE, MAE y Cross-Entropy
- Calcular Accuracy, Precision, Recall y F1
- Entender cuándo usar cada métrica
- Aplicar las funciones en modelos reales

## MSE (Mean Squared Error)

> MSE promedia los errores al cuadrado. Penaliza errores grandes más que pequeños. Fórmula: mean((actual - predicted)²).

```python
def mean_squared_error(actual, predicted):
    return sum((a - p) ** 2 for a, p in zip(actual, predicted)) / len(actual)

print(mean_squared_error([1, 2, 3], [1, 4, 2]))

print("\nCon NumPy:")
import numpy as np
actual = np.array([1, 2, 3])
predicted = np.array([1, 4, 2])
print(np.mean((actual - predicted) ** 2))
```

## MAE (Mean Absolute Error)

> MAE promedia los errores absolutos. Menos sensible a outliers que MSE. Fórmula: mean(|actual - predicted|).

```python
print("=== MAE ===")

def mean_absolute_error(actual, predicted):
    return sum(abs(a - p) for a, p in zip(actual, predicted)) / len(actual)

actual = np.array([1, 2, 3])
predicted = np.array([1, 4, 2])
print(f"MAE: {mean_absolute_error(actual, predicted):.4f}")

print("\nComparación MSE vs MAE:")
print("  MSE penaliza errores grandes más (errores^2)")
print("  MAE es más robusto a outliers")
```

## Cross-Entropy Loss

> Cross-entropy mide la diferencia entre dos distribuciones de probabilidad. Usada en clasificación. Para binary: -[y*log(p) + (1-y)*log(1-p)].

```python
import numpy as np

def binary_cross_entropy(y_true, y_pred):
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

y_true = np.array([1, 0, 1, 1, 0])
y_pred = np.array([0.9, 0.1, 0.8, 0.7, 0.2])

print(f"Binary Cross-Entropy: {binary_cross_entropy(y_true, y_pred):.4f}")

print("\nInterpretación:")
print("  Predicciones buenas → loss baja")
print("  Predicciones malas → loss alta")
```

## Matriz de Confusión

> La matriz de confusión muestra TP, TN, FP, FN. Es la base para calcular Accuracy, Precision, Recall y F1.

```python
print("=== Matriz de Confusión ===")

y_true = [1, 1, 0, 1, 0, 0, 1, 0, 1, 0]
y_pred = [1, 0, 0, 1, 0, 1, 1, 0, 0, 0]

tp = sum((t == 1 and p == 1) for t, p in zip(y_true, y_pred))
tn = sum((t == 0 and p == 0) for t, p in zip(y_true, y_pred))
fp = sum((t == 0 and p == 1) for t, p in zip(y_true, y_pred))
fn = sum((t == 1 and p == 0) for t, p in zip(y_true, y_pred))

print(f"Verdaderos Positivos (TP): {tp}")
print(f"Verdaderos Negativos (TN): {tn}")
print(f"Falsos Positivos (FP): {fp}")
print(f"Falsos Negativos (FN): {fn}")

print("\nMatriz de confusión:")
print(f"              Predicho 0  Predicho 1")
print(f"Real 0:          {tn}          {fp}")
print(f"Real 1:          {fn}          {tp}")
```

## Accuracy, Precision y Recall

> Accuracy = (TP+TN)/Total. Precision = TP/(TP+FP). Recall = TP/(TP+FN). Cada una mide algo diferente.

```python
print("=== Métricas de Clasificación ===")

tp, tn, fp, fn = 50, 30, 10, 10

accuracy = (tp + tn) / (tp + tn + fp + fn)
precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1 = 2 * precision * recall / (precision + recall)

print(f"Accuracy:  {accuracy:.4f} ({accuracy*100:.1f}% correctos)")
print(f"Precision: {precision:.4f} (de los predichos positivos, cuántos correctos)")
print(f"Recall:    {recall:.4f} (de los reales positivos, cuántos encontró)")
print(f"F1 Score:  {f1:.4f} (media armónica de precision y recall)")

print("\nCuándo usar cada una:")
print("  Accuracy: clases balanceadas, sin costo asimétrico")
print("  Precision: falsos positivos costosos (spam detection)")
print("  Recall: falsos negativos costosos (enfermedades)")
print("  F1: balance entre precision y recall")
```

## F1 Score

> F1 es la media armónica de Precision y Recall. Penaliza cuando una es alta y la otra baja. F1 = 2 * (P * R) / (P + R).

```python
print("=== F1 Score ===")

def f1_score(precision, recall):
    return 2 * precision * recall / (precision + recall)

print("Ejemplo: P=0.9, R=0.1")
print(f"  Media aritmética: {(0.9 + 0.1)/2:.2f} (engañosamente alta)")
print(f"  F1 Score: {f1_score(0.9, 0.1):.2f} (refleja el problema)")

print("\nPor qué media armónica?")
print("  Penaliza el desbalance. Si P=1 y R=0, F1=0")
print("  Media aritmética daría 0.5, lo cual es incorrecto")
```

## RMSE (Root Mean Squared Error)

> RMSE es la raíz cuadrada de MSE. Tiene las mismas unidades que los datos originales, lo que lo hace más interpretable.

```python
import numpy as np

actual = np.array([100, 200, 300, 400, 500])
predicted = np.array([110, 190, 320, 380, 510])

mse = np.mean((actual - predicted) ** 2)
rmse = np.sqrt(mse)
mae = np.mean(np.abs(actual - predicted))

print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f} (mismas unidades que los datos)")
print(f"MAE: {mae:.2f}")

print("\nInterpretación:")
print(f"  En promedio, las predicciones están a {rmse:.2f} unidades del valor real")
```

## Tips y Mejores Prácticas

> Para regresión: MSE si no hay outliers, MAE si los hay. RMSE es más interpretable que MSE.

> Para clasificación binaria: Binary Cross-Entropy. Para multi-clase: Categorical Cross-Entropy.

> Accuracy no funciona bien con clases desbalanceadas. Usa Precision, Recall, F1.

> F1 es mejor que Accuracy para datos desbalanceados porque solo considera la clase positiva.

> Cuando precision y recall son importantes, usa F1. Si una es más importante, ajusta los pesos.

## Errores Comunes

### Usar Accuracy en datos desbalanceados

¿Por qué ocurre?
- Si 95% son negativos, predecir siempre negativo da 95% accuracy.

Solución
- Usa F1, Precision o Recall según tu caso.

### MSE vs MAE en outliers

¿Por qué ocurre?
- MSE penaliza mucho un outlier (error²), MAE lo trata igual.

Solución
- Si outliers son ruido, usa MAE. Si son importantes, usa MSE.

### Log(0) en Cross-Entropy

¿Por qué ocurre?
- log(0) es infinito. Ocurre si y_pred=0 cuando y_true=1.

Solución
- Usa epsilon para clip: y_pred = clip(y_pred, epsilon, 1-epsilon).

### Mezclar loss con métricas

¿Por qué ocurre?
- Loss se usa para entrenar (diferenciable). Métricas para evaluar.

Solución
- Loss ≠ Métrica. Loss entrena el modelo; métricas evalúan el resultado final.
