## Media y Probabilidad

> La media es el promedio de los valores. La probabilidad empírica es la frecuencia relativa: count(target) / len(values). Para [1, 2, 2, 3] con target=2: media=2.0, probabilidad=2/4=0.5.

```python
def mean_and_probability(values, target):
    media = sum(values) / len(values)
    probabilidad = values.count(target) / len(values)
    return (media, probabilidad)

print(mean_and_probability([1, 2, 2, 3], 2))

print("\nCon NumPy:")
import numpy as np
arr = np.array([1, 2, 2, 3])
print(f"Media: {arr.mean()}")
print(f"Probabilidad de 2: {np.sum(arr == 2) / len(arr)}")
```

## Estadísticas Descriptivas Básicas

> Media, varianza y desviación estándar son la base para entender cualquier dataset. En ML las usas para normalizar, detectar outliers y comparar features.

```python
import numpy as np

puntajes = np.array([72, 85, 90, 78, 65, 88, 92, 70, 83, 75,
                     91, 68, 87, 74, 80, 95, 71, 84, 77, 89])

print(f"Media:    {puntajes.mean():.2f}")
print(f"Mediana:  {np.median(puntajes):.2f}")
print(f"Varianza: {puntajes.var():.2f}")
print(f"Std:      {puntajes.std():.2f}")
print(f"Rango:    {puntajes.max() - puntajes.min()}")
print(f"Q1: {np.percentile(puntajes, 25):.1f}")
print(f"Q3: {np.percentile(puntajes, 75):.1f}")
```

## Normalización y Estandarización

> Normalizar (0-1) o estandarizar (media=0, std=1) los datos mejora la convergencia del gradiente descendiente. Sin esto, features con rangos muy distintos dominan el entrenamiento.

```python
import numpy as np

edad     = np.array([25, 32, 41, 28, 55, 19, 37])
ingresos = np.array([30000, 55000, 80000, 42000, 120000, 25000, 65000])

def min_max(x):
    return (x - x.min()) / (x.max() - x.min())

def z_score(x):
    return (x - x.mean()) / x.std()

edad_norm    = min_max(edad)
ingresos_std = z_score(ingresos)

print("Edad normalizada (0-1):")
print(np.round(edad_norm, 3))
print("\nIngresos estandarizados:")
print(np.round(ingresos_std, 3))
print(f"\nMedia después: {ingresos_std.mean():.2f}, Std: {ingresos_std.std():.2f}")
```

## Correlación y Covarianza

> La correlación mide la relación lineal entre dos variables (-1 a 1). En feature engineering, features altamente correlacionadas son redundantes y pueden perjudicar modelos lineales.

```python
import numpy as np

horas   = np.array([2, 5, 1, 8, 3, 7, 4, 6, 9, 2])
puntaje = np.array([55, 78, 40, 92, 60, 88, 70, 85, 95, 50])

covarianza = np.cov(horas, puntaje)[0, 1]
correlacion = np.corrcoef(horas, puntaje)[0, 1]

print(f"Covarianza: {covarianza:.2f}")
print(f"Correlación de Pearson: {correlacion:.4f}")

if correlacion > 0.7:
    print("-> Correlación fuerte positiva")
elif correlacion < -0.7:
    print("-> Correlación fuerte negativa")
else:
    print("-> Correlación débil")
```

## Análisis de Distribución

> Analizar la distribución de un dataset permite decidir si normalizar o estandarizar. Si media y mediana difieren mucho, la distribución está sesgada.

```python
import numpy as np

np.random.seed(42)
precios = np.concatenate([
    np.random.normal(100, 20, 80),
    np.random.normal(300, 50, 20)
])

print("=== Análisis de distribución ===")
print(f"Media:   {precios.mean():.1f}")
print(f"Mediana: {np.median(precios):.1f}")
print(f"Std:     {precios.std():.1f}")
print(f"Sesgo (media-mediana): {precios.mean() - np.median(precios):.1f}")

z_scores = (precios - precios.mean()) / precios.std()
outliers = np.sum(np.abs(z_scores) > 3)
print(f"Outliers (|z| > 3): {outliers}")

p5, p95 = np.percentile(precios, [5, 95])
print(f"Rango 5%-95%: [{p5:.1f}, {p95:.1f}]")
```

## Probabilidad Básica

> Probabilidad condicional P(A|B) mide la probabilidad de A dado que B ocurrió. Bayes theorem: P(A|B) = P(B|A) * P(A) / P(B).

```python
print("=== Probabilidad Condicional ===")

print("Ejemplo: P(enfermo|positivo)")
print("  P(enfermo) = 0.01 (1% de la población)")
print("  P(positivo|enfermo) = 0.99 (99% de sensibilidad)")
print("  P(positivo|sano) = 0.05 (5% de falsos positivos)")

p_enfermo = 0.01
p_positivo_enfermo = 0.99
p_positivo_sano = 0.05

p_positivo = p_positivo_enfermo * p_enfermo + p_positivo_sano * (1 - p_enfermo)
p_enfermo_positivo = (p_positivo_enfermo * p_enfermo) / p_positivo

print(f"\nP(positivo) = {p_positivo:.4f}")
print(f"P(enfermo|positivo) = {p_enfermo_positivo:.4f}")
print(f"Aunque el test sea 99% sensible, solo {p_enfermo_positivo*100:.1f}% de positivos están enfermos")
```

## Tips y Mejores Prácticas

> Si media y mediana difieren mucho, la distribución está sesgada. Considera transformación logarítmica antes de entrenar.

> Z-score es preferida para redes neuronales y SVM. Min-Max es mejor cuando sabes los límites fijos (ej. pixels 0-255).

> Calcula media y std SOLO sobre el training set. Aplica esos valores al test set para evitar data leakage.

> La correlación solo mide relaciones lineales. Variables con relaciones no lineales pueden pasar desapercibidas.

## Errores Comunes

### Normalizar con estadísticas del test set

¿Por qué ocurre?
- El test set debe simular datos nuevos. Si usas sus estadísticas, el modelo ve el futuro.

Solución
- scaler.fit(X_train), luego scaler.transform(X_test).

### Ignorar la distribución antes de modelar

¿Por qué ocurre?
- Modelos lineales asumen distribución normal. Con datos muy sesgados los resultados son malos.

Solución
- Siempre haz análisis exploratorio: media, mediana, histograma, outliers.

### Confundir covarianza con correlación

¿Por qué ocurre?
- Covarianza no está normalizada, depende de las unidades. Correlación es adimensional [-1, 1].

Solución
- Usa correlación para comparar relaciones entre variables de cualquier escala.

### No detectar multicolinealidad

¿Por qué ocurre?
- Features altamente correlacionados causan problemas en regresión lineal.

Solución
- Calcula matriz de correlación antes de entrenar. Elimina o combina features redundantes.
