# Cálculo y Gradientes

Derivadas, regla de la cadena y gradientes — el motor detrás de backpropagation y la optimización

## Introducción

La derivada de f(x)=x² es f'(x)=2x. Usando diferencias centrales con h=0.001: (f(3+h) - f(3-h)) / (2h) ≈ (9.006 - 8.994) / 0.002 ≈ 6.0. Los gradientes son esenciales para backpropagation en redes neuronales.

### Objetivos de Aprendizaje

- Entender derivadas y la regla de la cadena
- Aproximar derivadas numéricamente
- Comprender gradientes en múltiples dimensiones
- Aplicar gradientes en optimización

## Gradiente Numérico

> La aproximación por diferencias centrales usa (f(x+h) - f(x-h)) / (2h). Para f(x)=x² en x=3: (3.001² - 2.999²) / 0.002 = 6.0

```python
def square_gradient(x, step):
    return ((x + step) ** 2 - (x - step) ** 2) / (2 * step)

print(square_gradient(3.0, 0.001))

print("\nDerivada analítica: f'(x) = 2x, f'(3) = 6")
print(f"Error: {abs(square_gradient(3.0, 0.001) - 6.0)}")
```

## Derivadas

> La derivada mide la tasa de cambio de una función. f'(x) indica cuánto cambia f cuando x cambia infinitesimalmente.

```python
print("=== Derivadas Básicas ===")

derivadas = {
    "f(x) = x^n": "f'(x) = n * x^(n-1)",
    "f(x) = e^x": "f'(x) = e^x",
    "f(x) = ln(x)": "f'(x) = 1/x",
    "f(x) = sin(x)": "f'(x) = cos(x)",
    "f(x) = cos(x)": "f'(x) = -sin(x)",
}

for funcion, derivada in derivadas.items():
    print(f"  {funcion} -> {derivada}")

print("\nEjemplo: f(x) = x^3, f'(x) = 3x^2")
print("  f'(2) = 3 * 4 = 12")
```

## Regla de la Cadena

> La regla de la cadena permite derivar funciones compuestas: si y = f(g(x)), entonces dy/dx = f'(g(x)) * g'(x).

```python
print("=== Regla de la Cadena ===")

print("Ejemplo: f(x) = (2x + 1)^3")
print("  Sea u = 2x + 1, entonces f = u^3")
print("  du/dx = 2, df/du = 3u^2")
print("  df/dx = df/du * du/dx = 3u^2 * 2 = 6u^2 = 6(2x+1)^2")

print("\nEn redes neuronales: sigmoid(z) donde z = x*w + b")
print("  d(sigmoid)/dz = sigmoid(z) * (1 - sigmoid(z))")
print("  dL/dw = dL/dsigmoid * dsigmoid/dz * dz/dw")

import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = 0.5
print(f"\nEjemplo numérico:")
print(f"  x = {x}")
print(f"  sigmoid(x) = {sigmoid(x):.4f}")
print(f"  d(sigmoid)/dx = {sigmoid(x) * (1 - sigmoid(x)):.4f}")
```

## Gradientes en ML

> Un gradiente es un vector de derivadas parciales. En ML, el gradiente indica la dirección de máximo ascenso de la función de pérdida.

```python
print("=== Gradientes en ML ===")

print("Función de pérdida: L(w) = (y - wx)^2")
print("Gradiente: dL/dw = -2x(y - wx)")

x, y, w = 2.0, 10.0, 3.0
loss = (y - w * x) ** 2
gradient = -2 * x * (y - w * x)

print(f"\n  x = {x}, y = {y}, w = {w}")
print(f"  Loss = {loss}")
print(f"  Gradiente dL/dw = {gradient}")

print("\nActualización de gradiente descendiente:")
learning_rate = 0.01
w_new = w - learning_rate * gradient
print(f"  w_new = w - lr * gradiente = {w_new:.4f}")
```

## Derivadas Parciales

> Para funciones de múltiples variables, las derivadas parciales miden el cambio respecto a cada variable manteniendo las demás constantes.

```python
print("=== Derivadas Parciales ===")

print("f(x, y) = x² + xy + y²")
print("  ∂f/∂x = 2x + y")
print("  ∂f/∂y = x + 2y")

x, y = 3.0, 4.0
df_dx = 2 * x + y
df_dy = x + 2 * y

print(f"\nEn (x=3, y=4):")
print(f"  ∂f/∂x = {df_dx}")
print(f"  ∂f/∂y = {df_dy}")
print(f"  Gradiente = ({df_dx}, {df_dy})")

print("\nMagnitud del gradiente:")
print(f"  ||gradiente|| = {np.sqrt(df_dx**2 + df_dy**2):.4f}")
```

## Gradiente Descendiente

> El algoritmo de gradiente descendiente minimiza una función moviéndose en la dirección opuesta al gradiente.

```python
print("=== Gradiente Descendiente ===")

def f(x):
    return x ** 2

def grad_f(x):
    return 2 * x

x = 10.0
lr = 0.1

print("Minimizar f(x) = x²")
print(f"  Punto inicial: x = {x}")

for i in range(10):
    x = x - lr * grad_f(x)
    print(f"  Iteración {i+1}: x = {x:.6f}, f(x) = {f(x):.6f}")

print("\nEl gradiente descendiente converge al mínimo x=0")
```

## Backpropagation

> Backpropagation usa la regla de la cadena para calcular gradientes capa por capa, desde la salida hacia la entrada.

```python
print("=== Backpropagation ===")

print("Red simple: input -> linear -> sigmoid -> output")
print("  z = x * w + b")
print("  y = sigmoid(z)")
print("  Loss = (y_target - y)²")

print("\nBackprop:")
print("  dL/dy = -2(y_target - y)")
print("  dy/dz = sigmoid(z)(1 - sigmoid(z))")
print("  dz/dw = x")
print("  dL/dw = dL/dy * dy/dz * dz/dw")

x, w, b, y_target = 2.0, 0.5, 0.1, 1.0
z = x * w + b
y = sigmoid(z)
loss = (y_target - y) ** 2

dL_dy = -2 * (y_target - y)
dy_dz = y * (1 - y)
dz_dw = x
dL_dw = dL_dy * dy_dz * dz_dw

print(f"\n  x={x}, w={w}, b={b}, y_target={y_target}")
print(f"  z={z:.4f}, y={y:.4f}, loss={loss:.4f}")
print(f"  dL/dw = {dL_dw:.4f}")
```

## Tips y Mejores Prácticas

> Diferencias centrales (f(x+h) - f(x-h)) / 2h es más preciso que diferencias hacia adelante o atrás.

> El gradiente indica la dirección de máximo ascenso. Para minimizar, mover en dirección opuesta.

> El learning rate controla qué tan rápido se actualizan los pesos. Muy alto = divergencia, muy bajo = convergencia lenta.

> La regla de la cadena es fundamental: multiplica gradientes capa por capa en backprop.

> En ML, el gradiente se usa para actualizar pesos: w = w - lr * dL/dw.

## Errores Comunes

### Olvidar la regla de la cadena

¿Por qué ocurre?
- Se olvida multiplicar los gradientes intermedios.

Solución
- En funciones compuestas, SIEMPRE aplicar regla de la cadena.

### Learning rate muy grande

¿Por qué ocurre?
- Actualizaciones grandes causan divergencia.

Solución
- Reducir learning rate o usar técnicas como Adam.

### Confundir gradiente con derivada

¿Por qué ocurre?
- En una variable son iguales. En múltiples variables, gradiente = vector de derivadas parciales.

Solución
- Gradiente es un vector. Derivada puede ser un escalar (1D) o parte del gradiente (nD).

### Vanishing gradients

¿Por qué ocurre?
- En redes profundas, productos de gradientes < 1 hacen que el gradiente se acerque a 0.

Solución
- Usar ReLU, skip connections, o normalización de capas.
