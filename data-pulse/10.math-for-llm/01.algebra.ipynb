# Álgebra Lineal para ML

Vectores, matrices y multiplicación matricial — el lenguaje de embeddings, transformaciones y pesos

## Introducción

El producto punto multiplica pares de valores correspondientes y suma los resultados. Es la base para embeddings,相似idad de vectores, y el cálculo de predicciones en redes neuronales.

### Objetivos de Aprendizaje

- Entender vectores y matrices como representaciones de datos
- Calcular el producto punto entre vectores
- Realizar multiplicación de matrices
- Aplicar algebra lineal en Machine Learning

## Producto Punto

> El producto punto multiplica pares de valores correspondientes y suma los resultados. Para [1, 2, 3] · [4, 5, 6] = 1×4 + 2×5 + 3×6 = 32.

```python
def dot_product(left, right):
    return sum(a * b for a, b in zip(left, right))

print(dot_product([1, 2, 3], [4, 5, 6]))

print("\nCon NumPy:")
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.dot(a, b))
print(a @ b)
```

## Vectores

> Un vector es una lista ordenada de números. En ML, los vectores representan embeddings de palabras, características de datos, o direcciones en el espacio.

```python
print("=== Vectores ===")

import numpy as np

v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

print(f"v1 = {v1}")
print(f"v2 = {v2}")

print("\nOperaciones:")
print(f"  Suma: v1 + v2 = {v1 + v2}")
print(f"  Resta: v1 - v2 = {v1 - v2}")
print(f"  Escalar * 2: {v1 * 2}")

print("\nMagnitud (norma L2):")
print(f"  ||v1|| = {np.linalg.norm(v1)}")
```

## Matrices

> Una matriz es un array 2D de números. En ML, las filas son típicamente muestras y las columnas son features.

```python
print("=== Matrices ===")

A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[7, 8], [9, 10], [11, 12]])

print(f"Matriz A (2x3):\n{A}")
print(f"\nMatriz B (3x2):\n{B}")

print(f"\nShape de A: {A.shape}")
print(f"Shape de B: {B.shape}")

print("\nTranspuesta de A:")
print(A.T)
```

## Multiplicación de Matrices

> Para multiplicar matrices A @ B, el número de columnas de A debe igualar el número de filas de B. El elemento (i,j) resulta del producto punto de la fila i de A con la columna j de B.

```python
print("=== Multiplicación de Matrices ===")

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(f"A:\n{A}")
print(f"\nB:\n{B}")
print(f"\nA @ B:\n{A @ B}")

print("\nExplicación del cálculo:")
print("  (A @ B)[0,0] = fila0(A) · columna0(B) = 1*5 + 2*7 = 19")
print("  (A @ B)[0,1] = fila0(A) · columna1(B) = 1*6 + 2*8 = 22")
print("  (A @ B)[1,0] = fila1(A) · columna0(B) = 3*5 + 4*7 = 43")
print("  (A @ B)[1,1] = fila1(A) · columna1(B) = 3*6 + 4*8 = 50")
```

## Embeddings en ML

> Los embeddings son vectores densos que representan texto, imágenes o audio. Palabras similares tienen embeddings cercanos en el espacio vectorial.

```python
print("=== Embeddings ===")

palabras = {
    "king": np.array([0.9, 0.3, 0.2]),
    "queen": np.array([0.85, 0.35, 0.25]),
    "man": np.array([0.7, 0.8, 0.5]),
    "woman": np.array([0.75, 0.75, 0.55]),
}

print("Embeddings simplificados (3 dimensiones):")
for palabra, embedding in palabras.items():
    print(f"  {palabra}: {embedding}")

print("\nSimilitud coseno entre king y queen:")
v1, v2 = palabras["king"], palabras["queen"]
cos_sim = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
print(f"  {cos_sim:.4f}")

print("\nking - man + woman ≈ queen (algebra lineal!)")
resultado = palabras["king"] - palabras["man"] + palabras["woman"]
print(f"  {resultado}")
```

## Pesos de Redes Neuronales

> Los pesos de una red neuronal son matrices. La propagación hacia adelante es multiplicación de matrices: output = input @ weights + bias.

```python
print("=== Pesos de Red Neuronal ===")

np.random.seed(42)

input_features = np.array([1.0, 0.5, 0.3])
weights = np.random.randn(3, 4)
bias = np.array([0.1, 0.2, 0.3, 0.4])

print(f"Input (3 features): {input_features}")
print(f"\nWeights (3 x 4):\n{weights}")
print(f"\nBias (4 neuronios): {bias}")

output = input_features @ weights + bias
print(f"\nOutput (4 neuronios): {output}")

print("\nExplicación:")
print("  output[j] = sum(input[i] * weights[i,j]) + bias[j]")
print("  output[0] = sum([1.0*w[0,0], 0.5*w[1,0], 0.3*w[2,0]]) + 0.1")
```

## Transformaciones Lineales

> Las matrices pueden representar transformaciones: rotación, escalado,剪切. Multiplicar un vector por una matriz lo transforma.

```python
print("=== Transformaciones ===")

import numpy as np

v = np.array([1, 0])
print(f"Vector original: {v}")

theta = np.pi / 4
rotacion = np.array([[np.cos(theta), -np.sin(theta)],
                     [np.sin(theta), np.cos(theta)]])

v_rotado = rotacion @ v
print(f"\nRotación 45 grados: {v_rotado}")

escala = np.array([[2, 0], [0, 2]])
v_escalado = escala @ v
print(f"\nEscalado 2x: {v_escalado}")
```

## Tips y Mejores Prácticas

> El producto punto es la base de similarity entre vectores. Usa np.dot() o @.

> Para similitud coseno: cos_sim = np.dot(a, b) / (||a|| * ||b||).

> Los embeddings permiten operaciones algebraicas: king - man + woman ≈ queen.

> En redes neuronales, forward pass = input @ weights + bias.

> Verifica shapes antes de multiplicar: (m,n) @ (n,p) = (m,p).

## Errores Comunes

### Dimensiones incompatibles en producto punto

¿Por qué ocurre?
- Intentar dot(a, b) cuando len(a) != len(b).

Solución
- Verifica que ambos vectores tengan la misma longitud.

### Confundir * con @

¿Por qué ocurre?
- * multiplica elemento a elemento, no hace producto punto.

Solución
- Usa @ o np.dot() para producto punto matricial.

### Shapes equivocadas en multiplicación de matrices

¿Por qué ocurre?
- Intentar (2,3) @ (2,3) cuando debe ser (2,3) @ (3,2).

Solución
- Recuerda: columnas de A = filas de B.
