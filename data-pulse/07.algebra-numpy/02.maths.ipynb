```python
# Ventas mensuales: 6 meses x 4 regiones
np.random.seed(1)
ventas = np.random.randint(50, 200, size=(6, 4))
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun']
regiones = ['Norte', 'Sur', 'Este', 'Oeste']

print(ventas)

print("\nMinimal statistics")
for i, region in enumerate(regiones):
    sells = ventas[:, i]
    print(f"reg: {region}, mean: {sells.mean()}, total: {sells.sum()}, min-max: {sells.min()} - {sells.max()}")
    
print("\nBy month")
sells = ventas.sum(axis=1)
max_sells_i = np.argmax(sells)
print(sells, f"month: {meses[max_sells_i]}, max: {sells[max_sells_i]}")

print("\nGrow month by month")
grow = np.diff(sells)
print(sells, grow)
```

```python
def add_vectors(left, right):
    result = []
    for i in range(0, len(left)):
        result.append(left[i] + right[i])
        
    return result

print(add_vectors([1, 2, 3], [4, 5, 6]))
```

```python
def inner_values(values):
    return values[1:-1]

print(inner_values([10, 20, 30, 40]))
```
